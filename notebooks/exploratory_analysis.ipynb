# AI-Powered Water Sustainability Assistant — Exploratory Data Analysis & Modeling

**SDG 6: Clean Water and Sanitation**

This notebook demonstrates the complete analytical workflow:
1. Data Loading & Schema Cleaning
2. Descriptive Statistics & Activity Distribution
3. Anomaly Detection (Z-Score & IQR)
4. Time-Series Forecasting Baseline
5. Personal Water Sustainability Index Calculation

```python
import pandas as pd
import numpy as np

# 1. Load synthetic demo dataset
df = pd.read_csv('../data/demo/sample_water_consumption.csv')
df['date'] = pd.to_datetime(df['date'])
print(f"Loaded {len(df)} daily water consumption records.")
df.head()
```

```python
# 2. Calculate Total Daily Volume & Descriptive Stats
activity_cols = [c for c in df.columns if c != 'date']
df['total'] = df[activity_cols].sum(axis=1)

stats = {
    'Total Consumption (L)': df['total'].sum(),
    'Mean Daily (L)': df['total'].mean(),
    'Median Daily (L)': df['total'].median(),
    'Min Daily (L)': df['total'].min(),
    'Max Daily (L)': df['total'].max(),
    'Std Dev (L)': df['total'].std()
}
pd.Series(stats)
```

```python
# 3. Activity Distribution Percentages
act_totals = df[activity_cols].sum()
act_pcts = (act_totals / df['total'].sum() * 100).round(1)
activity_df = pd.DataFrame({'Total Liters': act_totals, 'Percentage Share (%)': act_pcts})
activity_df.sort_values(by='Percentage Share (%)', ascending=False)
```

```python
# 4. Anomaly Detection (Z-Score Method)
mean_val = df['total'].mean()
std_val = df['total'].std()
df['z_score'] = (df['total'] - mean_val) / std_val

anomalies = df[df['z_score'] >= 2.0]
print(f"Flagged {len(anomalies)} anomaly days:")
anomalies[['date', 'total', 'z_score']]
```

```python
# 5. Linear Time-Series Forecasting Baseline
from sklearn.linear_model import LinearRegression

df['time_idx'] = np.arange(len(df))
df['day_of_week'] = df['date'].dt.dayofweek

X = df[['time_idx', 'day_of_week']]
y = df['total']

model = LinearRegression()
model.fit(X, y)

print(f"Forecast Model trained. R^2 score: {model.score(X, y):.3f}")
```
